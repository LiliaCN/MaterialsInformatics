# Check that libraries are installed

This notebook checks to see if you have the correct version of Python as well as all necessary libraries installed.

Check the [main README file](https://github.com/anthony-wang/BestPractices/blob/master/README.md) for instructions if anything is missing.

```python
import sys
from importlib.metadata import version, PackageNotFoundError
from packaging.version import Version

OK, FAIL = "[ OK ]", "[FAIL]"

print("Using Python in", sys.prefix)
print(sys.version)
if Version(sys.version.split()[0]) < Version("3.11"):
    print(FAIL, "Python >= 3.11 is required.")
print()

# distribution (PyPI) name -> minimum acceptable version
requirements = {
    "numpy": "1.26",
    "pandas": "2.0",
    "matplotlib": "3.8",
    "seaborn": "0.12",
    "scikit-learn": "1.4",
    "scipy": "1.11",
    "tqdm": "4.64",
    "torch": "2.5",
    "cbfv": "1.1",
    "ydata-profiling": "4.12",
}

for dist, min_ver in requirements.items():
    try:
        installed = version(dist)
        status = OK if Version(installed) >= Version(min_ver) else FAIL
        print(f"{status} {dist} {installed}  (need >= {min_ver})")
    except PackageNotFoundError:
        print(FAIL, f"{dist} is NOT installed")
```
